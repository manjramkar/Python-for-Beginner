```python
#20th Que: A year is a leap year if it is divisible by 4, but not by 100 unless it is also divisible by 400.

Year = int(input("Enter year="))

if(Year % 4 == 0 and Year % 100 !=0) or (Year % 400 == 0):
    print("It is leap year=", Year)
else:
    print("It's not leap year=", Year)
```
