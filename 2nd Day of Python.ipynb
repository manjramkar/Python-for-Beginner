```python
# String Functions

str= "my name is Akshay Manjramkar"
print(str.endswith("kar")) #return true if string ends with substr
print(str.capitalize()) #capitalizes 1st char
print(str.replace("a","f")) #replaces all occcurrences of old
print(str.find("a")) # return 1st index of 1st occurrer
print(str.find("q")) # return 1st index of 1st occurrer in negative
print(str.count("s")) # counts the occurrences of substr
```

```python
#Conditional Statements

# if-elif-else(syntax)

#if Condition:
age= 24
if(age>=18):
    print("can vote")
    print("can drive")
```

```python
#elif condition
light = "yellow"

if(light == "red"):
    print("stop")
elif(light == "green"):
    print("go")
elif(light == "yellow"):
    print("look")
```

```python
#else condition

light = "blue"

if(light == "red"):
    print("stop")
elif(light == "green"):
    print("go")
elif(light == "yellow"):
    print("look")

else:
    print("light is broken")
```

```python
# Grade students bsed on their marks

marks = int(input("enter student marks: "))

if(marks >= 90):
    grade = "A"

elif(marks >= 80 and marks < 90):
    grade = "B"
elif(marks >= 70 and marks < 80):
    grade = "C"
else:
    grade = "D"
print("grade of the student ->" , grade)
```

```python
#nesting (you can add another sentense)

age = 17

if(age >= 18):
    if(age >= 80):
        print("cannot drive")
    else:
        print("can drive")
        
else:
    print("cannot drive")
```
